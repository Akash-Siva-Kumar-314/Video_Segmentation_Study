```python
import ultralytics
ultralytics.checks()
from ultralytics import YOLO
```

```python
# Load a model
# You can use different YOLOv8 variants (yolov8n-seg, yolov8s-seg, yolov8m-seg, yolov8l-seg, yolov8nx-seg)
model = YOLO('yolov8n-seg.pt')  # load a pretrained model
```

```python
print(model.names.values())
```

```python
'''
YOLOv8 Segmentation models are pretrained on COCO dataset with the following Classes.
['person', 'bicycle'.... 'cat', 'dog', 'horse', 'sheep', 'cow'... 'backpack', 'umbrella', 'handbag', 'tie', 'suitcase'... etc etc]
Cow is a part of that dataset!
'''
```

```python
# Lets use the model to predict on an image
results = model('/content/drive/MyDrive/Video_Segmentation_Study/Video_Segmentation_Study01/Cow.jpg')
```

```python
results[0].save('/content/drive/MyDrive/Video_Segmentation_Study/Video_Segmentation_Study01/Results/output.jpg')
```
